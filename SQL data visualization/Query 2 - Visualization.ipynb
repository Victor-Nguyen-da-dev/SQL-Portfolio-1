```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import adjustText as adjust_text
import ast

df = pd.read_csv('C:\Test\Github\SQL\SQL data visualization\Query 2.csv')
```

```python
df2 = df.groupby('skills').size().sort_values(ascending=False).head(10).copy()
```

```python
df2 = df2.reset_index(name='skill_count')
df2
```

```python
sns.barplot(data=df2, x='skill_count', y='skills', hue='skill_count', palette='Blues', legend=False, edgecolor='black')
sns.set_theme(style='darkgrid')

plt.title('Skill Counts of the Top 10 Paying Data Analyst Jobs in 2023')
plt.xlabel('Frequency')
plt.ylabel('')
plt.show()
```
